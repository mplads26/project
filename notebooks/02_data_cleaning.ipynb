```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd().parent

lok_path = BASE_DIR / "data" / "raw" / "MPLADS_Lok_Sabha.xlsx"
rajya_path = BASE_DIR / "data" / "raw" / "MPLADS_Rajya_Sabha.xlsx"

lok_df = pd.read_excel(lok_path)
rajya_df = pd.read_excel(rajya_path)
```

```python
lok_clean = lok_df[lok_df["Sr. No."] != "Grand Total"].copy()
rajya_clean = rajya_df[rajya_df["Sr. No."] != "Grand Total"].copy()

print("Lok Sabha before:", lok_df.shape)
print("Lok Sabha after:", lok_clean.shape)

print("Rajya Sabha before:", rajya_df.shape)
print("Rajya Sabha after:", rajya_clean.shape)
```

```python
print("Missing values in Lok Sabha:")
print(lok_clean.isnull().sum())

print("\nMissing values in Rajya Sabha:")
print(rajya_clean.isnull().sum())
```

```python
missing_lok = lok_clean[lok_clean["Allocated AMOUNT ( ₹ )"].isnull()]

print("Rows with missing allocation:")
display(missing_lok)
```

```python
# Rename Lok Sabha columns
lok_clean = lok_clean.rename(columns={
    "Sr. No.": "sr_no",
    "State": "state",
    "Hon'ble Members of Parliaments": "mp_name",
    "Constituency": "constituency",
    "Allocated AMOUNT ( ₹ )": "allocated_amount"
})

# Rename Rajya Sabha columns
rajya_clean = rajya_clean.rename(columns={
    "Sr. No.": "sr_no",
    "State": "state",
    "Hon'ble Members of Parliament": "mp_name",
    "Elected/Nominated": "member_type",
    "Allocated AMOUNT ( ₹ )": "allocated_amount"
})

print("Lok Sabha columns:")
print(lok_clean.columns.tolist())

print("\nRajya Sabha columns:")
print(rajya_clean.columns.tolist())
```

```python
print("Lok Sabha allocated_amount dtype:")
print(lok_clean["allocated_amount"].dtype)

print("\nLok Sabha sample values:")
print(lok_clean["allocated_amount"].head(10).tolist())

print("\nRajya Sabha allocated_amount dtype:")
print(rajya_clean["allocated_amount"].dtype)

print("\nRajya Sabha sample values:")
print(rajya_clean["allocated_amount"].head(10).tolist())
```

```python
# Convert allocation amounts to numeric values
lok_clean["allocated_amount"] = pd.to_numeric(
    lok_clean["allocated_amount"],
    errors="coerce"
)

rajya_clean["allocated_amount"] = pd.to_numeric(
    rajya_clean["allocated_amount"],
    errors="coerce"
)

print("Lok Sabha dtype:", lok_clean["allocated_amount"].dtype)
print("Rajya Sabha dtype:", rajya_clean["allocated_amount"].dtype)
```

```python
print("Missing allocated amounts after conversion:")

print("Lok Sabha:", lok_clean["allocated_amount"].isnull().sum())
print("Rajya Sabha:", rajya_clean["allocated_amount"].isnull().sum())
```

```python
# Clean text columns in Lok Sabha
lok_text_cols = ["state", "mp_name", "constituency"]

for col in lok_text_cols:
    lok_clean[col] = lok_clean[col].astype(str).str.strip()

# Clean text columns in Rajya Sabha
rajya_text_cols = ["state", "mp_name", "member_type"]

for col in rajya_text_cols:
    rajya_clean[col] = rajya_clean[col].astype(str).str.strip()

print("Text columns cleaned successfully.")
```

```python
print(rajya_clean["member_type"].value_counts())
```

```python
print("Lok Sabha states:")
print(sorted(lok_clean["state"].unique()))

print("\nRajya Sabha states:")
print(sorted(rajya_clean["state"].unique()))
```

```python
# Convert serial numbers to numeric
lok_clean["sr_no"] = pd.to_numeric(lok_clean["sr_no"], errors="coerce")
rajya_clean["sr_no"] = pd.to_numeric(rajya_clean["sr_no"], errors="coerce")

print("Lok Sabha sr_no dtype:", lok_clean["sr_no"].dtype)
print("Rajya Sabha sr_no dtype:", rajya_clean["sr_no"].dtype)

print("\nMissing sr_no values:")
print("Lok Sabha:", lok_clean["sr_no"].isnull().sum())
print("Rajya Sabha:", rajya_clean["sr_no"].isnull().sum())
```

```python
print("===== LOK SABHA FINAL CHECK =====")
print("Shape:", lok_clean.shape)
print("\nData types:")
print(lok_clean.dtypes)
print("\nMissing values:")
print(lok_clean.isnull().sum())
print("\nDuplicate rows:", lok_clean.duplicated().sum())

print("\n" + "="*40)

print("===== RAJYA SABHA FINAL CHECK =====")
print("Shape:", rajya_clean.shape)
print("\nData types:")
print(rajya_clean.dtypes)
print("\nMissing values:")
print(rajya_clean.isnull().sum())
print("\nDuplicate rows:", rajya_clean.duplicated().sum())
```

```python
processed_dir = BASE_DIR / "data" / "processed"

# Save Lok Sabha cleaned data
lok_clean.to_csv(
    processed_dir / "MPLADS_Lok_Sabha_cleaned.csv",
    index=False
)

# Save Rajya Sabha cleaned data
rajya_clean.to_csv(
    processed_dir / "MPLADS_Rajya_Sabha_cleaned.csv",
    index=False
)

print("Cleaned files saved successfully!")
```
