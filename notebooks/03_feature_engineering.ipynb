```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd().parent
processed_dir = BASE_DIR / "data" / "processed"

lok_df = pd.read_csv(
    processed_dir / "MPLADS_Lok_Sabha_cleaned.csv"
)

rajya_df = pd.read_csv(
    processed_dir / "MPLADS_Rajya_Sabha_cleaned.csv"
)

print("Lok Sabha:", lok_df.shape)
print("Rajya Sabha:", rajya_df.shape)
```

```python
print(lok_df.head())
print(rajya_df.head())
```

```python
print("LOK SABHA ALLOCATION STATISTICS")
print(lok_df["allocated_amount"].describe())

print("\nRAJYA SABHA ALLOCATION STATISTICS")
print(rajya_df["allocated_amount"].describe())
```

```python
lok_df["house"] = "Lok Sabha"
rajya_df["house"] = "Rajya Sabha"

print(lok_df[["mp_name", "house"]].head())
print()
print(rajya_df[["mp_name", "house"]].head())
```

```python
lok_mean = lok_df["allocated_amount"].mean()
rajya_mean = rajya_df["allocated_amount"].mean()

lok_df["allocation_vs_mean"] = (
    lok_df["allocated_amount"] - lok_mean
)

rajya_df["allocation_vs_mean"] = (
    rajya_df["allocated_amount"] - rajya_mean
)
```

```python
lok_df["allocation_pct_from_mean"] = (
    (lok_df["allocated_amount"] - lok_mean) / lok_mean
) * 100

rajya_df["allocation_pct_from_mean"] = (
    (rajya_df["allocated_amount"] - rajya_mean) / rajya_mean
) * 100
```

```python
print("LOK SABHA")
display(
    lok_df[
        [
            "mp_name",
            "state",
            "allocated_amount",
            "allocation_vs_mean",
            "allocation_pct_from_mean"
        ]
    ].head()
)

print("\nRAJYA SABHA")
display(
    rajya_df[
        [
            "mp_name",
            "state",
            "allocated_amount",
            "allocation_vs_mean",
            "allocation_pct_from_mean"
        ]
    ].head()
)
```

```python
# State-wise average allocation

lok_state_mean = (
    lok_df.groupby("state")["allocated_amount"]
    .transform("mean")
)

rajya_state_mean = (
    rajya_df.groupby("state")["allocated_amount"]
    .transform("mean")
)

lok_df["state_avg_allocation"] = lok_state_mean
rajya_df["state_avg_allocation"] = rajya_state_mean
```

```python
lok_df["allocation_vs_state_avg"] = (
    lok_df["allocated_amount"] - lok_df["state_avg_allocation"]
)

rajya_df["allocation_vs_state_avg"] = (
    rajya_df["allocated_amount"] - rajya_df["state_avg_allocation"]
)
```

```python
lok_df["allocation_pct_from_state_avg"] = (
    lok_df["allocation_vs_state_avg"]
    / lok_df["state_avg_allocation"]
) * 100

rajya_df["allocation_pct_from_state_avg"] = (
    rajya_df["allocation_vs_state_avg"]
    / rajya_df["state_avg_allocation"]
) * 100
```

```python
print("LOK SABHA")
display(
    lok_df[
        [
            "mp_name",
            "state",
            "allocated_amount",
            "state_avg_allocation",
            "allocation_vs_state_avg",
            "allocation_pct_from_state_avg"
        ]
    ].head(10)
)

print("\nRAJYA SABHA")
display(
    rajya_df[
        [
            "mp_name",
            "state",
            "allocated_amount",
            "state_avg_allocation",
            "allocation_vs_state_avg",
            "allocation_pct_from_state_avg"
        ]
    ].head(10)
)
```

```python
# Calculate allocation Z-scores

lok_std = lok_df["allocated_amount"].std()
rajya_std = rajya_df["allocated_amount"].std()

lok_df["allocation_zscore"] = (
    (lok_df["allocated_amount"] - lok_mean) / lok_std
)

rajya_df["allocation_zscore"] = (
    (rajya_df["allocated_amount"] - rajya_mean) / rajya_std
)

print("Z-score features created successfully.")
```

```python
print("===== TOP 10 MOST UNUSUAL LOK SABHA ALLOCATIONS =====")

display(
    lok_df.loc[
        lok_df["allocation_zscore"].abs().sort_values(ascending=False).index,
        ["mp_name", "state", "allocated_amount", "allocation_zscore"]
    ].head(10)
)

print("===== TOP 10 MOST UNUSUAL RAJYA SABHA ALLOCATIONS =====")

display(
    rajya_df.loc[
        rajya_df["allocation_zscore"].abs().sort_values(ascending=False).index,
        ["mp_name", "state", "allocated_amount", "allocation_zscore"]
    ].head(10)
)
```

```python
# Create Z-score based anomaly flags

lok_df["zscore_anomaly"] = (
    lok_df["allocation_zscore"].abs() > 3
)

rajya_df["zscore_anomaly"] = (
    rajya_df["allocation_zscore"].abs() > 3
)

print("Lok Sabha anomalies:", lok_df["zscore_anomaly"].sum())
print("Rajya Sabha anomalies:", rajya_df["zscore_anomaly"].sum())
```

```python
print("===== FLAGGED LOK SABHA RECORDS =====")

display(
    lok_df[
        lok_df["zscore_anomaly"]
    ][
        [
            "mp_name",
            "state",
            "allocated_amount",
            "allocation_zscore"
        ]
    ].sort_values(
        "allocation_zscore",
        ascending=False
    )
)

print("===== FLAGGED RAJYA SABHA RECORDS =====")

display(
    rajya_df[
        rajya_df["zscore_anomaly"]
    ][
        [
            "mp_name",
            "state",
            "allocated_amount",
            "allocation_zscore"
        ]
    ].sort_values(
        "allocation_zscore",
        ascending=False
    )
)
```

```python
def assign_risk_level(z):
    if pd.isna(z):
        return "Unknown"
    elif abs(z) > 3:
        return "High"
    elif abs(z) > 2:
        return "Medium"
    else:
        return "Low"


lok_df["risk_level"] = lok_df["allocation_zscore"].apply(assign_risk_level)
rajya_df["risk_level"] = rajya_df["allocation_zscore"].apply(assign_risk_level)

print("Lok Sabha risk levels:")
print(lok_df["risk_level"].value_counts())

print("\nRajya Sabha risk levels:")
print(rajya_df["risk_level"].value_counts())
```

```python
print("===== HIGH-RISK LOK SABHA RECORDS =====")

display(
    lok_df[lok_df["risk_level"] == "High"][
        [
            "mp_name",
            "state",
            "allocated_amount",
            "allocation_zscore",
            "risk_level"
        ]
    ].sort_values("allocation_zscore", ascending=False)
)

print("===== HIGH-RISK RAJYA SABHA RECORDS =====")

display(
    rajya_df[rajya_df["risk_level"] == "High"][
        [
            "mp_name",
            "state",
            "allocated_amount",
            "allocation_zscore",
            "risk_level"
        ]
    ].sort_values("allocation_zscore", ascending=False)
)
```

```python
def get_anomaly_reason(z):
    if pd.isna(z):
        return "Allocation amount is missing"

    elif z > 3:
        return "Allocation significantly above the normal range"

    elif z < -3:
        return "Allocation significantly below the normal range"

    elif z > 2:
        return "Allocation moderately above the normal range"

    elif z < -2:
        return "Allocation moderately below the normal range"

    else:
        return "Allocation within the normal range"


lok_df["anomaly_reason"] = lok_df["allocation_zscore"].apply(
    get_anomaly_reason
)

rajya_df["anomaly_reason"] = rajya_df["allocation_zscore"].apply(
    get_anomaly_reason
)

print("Explainable anomaly reasons added successfully.")
```

```python
print("===== LOK SABHA EXPLANATION CHECK =====")

display(
    lok_df[
        [
            "mp_name",
            "allocated_amount",
            "allocation_zscore",
            "risk_level",
            "anomaly_reason"
        ]
    ].sort_values(
        "allocation_zscore",
        ascending=False
    ).head(15)
)
```
