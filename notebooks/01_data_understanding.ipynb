```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd().parent

lok_path = BASE_DIR / "data" / "raw" / "MPLADS_Lok_Sabha.xlsx"
rajya_path = BASE_DIR / "data" / "raw" / "MPLADS_Rajya_Sabha.xlsx"

lok_df = pd.read_excel(lok_path)
rajya_df = pd.read_excel(rajya_path)

print("Lok Sabha shape:", lok_df.shape)
print("Rajya Sabha shape:", rajya_df.shape)
```

```python
print("LOK SABHA COLUMNS:")
print(lok_df.columns.tolist())

print("\nRAJYA SABHA COLUMNS:")
print(rajya_df.columns.tolist())
```

```python
print("LOK SABHA - FIRST 5 ROWS")
display(lok_df.head())

print("\nRAJYA SABHA - FIRST 5 ROWS")
display(rajya_df.head())
```

```python
print("===== LOK SABHA INFO =====")
lok_df.info()

print("\n===== RAJYA SABHA INFO =====")
rajya_df.info()
```

```python
print("===== LOK SABHA MISSING VALUES =====")
print(lok_df.isnull().sum())

print("\n===== RAJYA SABHA MISSING VALUES =====")
print(rajya_df.isnull().sum())
```

```python
print("LOK SABHA - ROW WITH MISSING ALLOCATION:")
display(lok_df[lok_df["Allocated AMOUNT ( ₹ )"].isnull()])
```

```python
print("Lok Sabha exact duplicate rows:", lok_df.duplicated().sum())
print("Rajya Sabha exact duplicate rows:", rajya_df.duplicated().sum())
```
